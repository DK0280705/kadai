```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

num = int(input())
print(is_prime(num))
```

```python
date = input()
# the date input is mmdd, split it into month and day
month = int(date[:2])
day = int(date[2:])
# Count the number of days from the start of the year to the given date
days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
total_days = sum(days_in_month[:month-1]) + day
# Starting at sunday on day 1, find the day of the week for the given date
days = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
print(total_days, days[(total_days - 1) % 7])
```

```python
year = int(input())
month = int(input())
day = int(input())

def is_date_valid(year, month, day):
    if month < 1 or month > 12:
        return False
    days_in_month = [31, 28 + (1 if (year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)) else 0), 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if day < 1 or day > days_in_month[month - 1]:
        return False
    return True

print(is_date_valid(year, month, day))
```
